```python
from datasets import load_dataset

dataset = load_dataset(
    "ai4bharat/BPCC",
    "ilci",
    split="hin_Deva"
)

print(dataset)
print("Columns:", dataset.column_names)
print("First example:", dataset[0])
```

```python

import pandas as pd
import numpy as np
from pathlib import Path

SEED = 42
OUT = Path("mt_data")
OUT.mkdir(parents=True, exist_ok=True)

# Convert the Hugging Face dataset to a DataFrame.
df = dataset.to_pandas()[["src_lang", "tgt_lang", "src", "tgt"]]

# Keep only English-to-Hindi pairs.
df = df[
    (df["src_lang"] == "eng_Latn") &
    (df["tgt_lang"] == "hin_Deva")
].copy()

df = df.rename(columns={"src": "en", "tgt": "hi"})
df = df[["en", "hi"]].dropna()

df["en"] = df["en"].astype(str).str.strip()
df["hi"] = df["hi"].astype(str).str.strip()
df = df[(df["en"] != "") & (df["hi"] != "")]

# Remove exact duplicate parallel pairs.
df = df.drop_duplicates(subset=["en", "hi"]).reset_index(drop=True)

# Prevent identical English sources from appearing in multiple splits.
df = df.drop_duplicates(subset=["en"]).reset_index(drop=True)

# Bin sentences by English whitespace-separated word count.
df["src_length"] = df["en"].str.split().str.len()

def assign_bin(n):
    if n <= 10:
        return "bin1"
    elif n < 20:
        return "bin2"
    elif n < 30:
        return "bin3"
    return "bin4"

df["bin"] = df["src_length"].apply(assign_bin)

# Split each bin separately: 80% train, 10% validation, 10% test.
train_parts, valid_parts, test_parts = [], [], []

for bin_name in ["bin1", "bin2", "bin3", "bin4"]:
    group = df[df["bin"] == bin_name].sample(
        frac=1, random_state=SEED
    ).reset_index(drop=True)

    n = len(group)
    n_test = max(1, round(n * 0.10))
    n_valid = max(1, round(n * 0.10))

    test_parts.append(group.iloc[:n_test])
    valid_parts.append(group.iloc[n_test:n_test + n_valid])
    train_parts.append(group.iloc[n_test + n_valid:])

train_df = pd.concat(train_parts).sample(
    frac=1, random_state=SEED
).reset_index(drop=True)

valid_df = pd.concat(valid_parts).sample(
    frac=1, random_state=SEED
).reset_index(drop=True)

test_df = pd.concat(test_parts).sample(
    frac=1, random_state=SEED
).reset_index(drop=True)

# Write aligned parallel files.
for name, frame in [
    ("train", train_df),
    ("valid", valid_df),
    ("test", test_df),
]:
    frame["en"].to_csv(
        OUT / f"{name}.en", index=False, header=False
    )
    frame["hi"].to_csv(
        OUT / f"{name}.hi", index=False, header=False
    )

# Preserve test metadata for bin-wise BLEU evaluation.
test_df[["en", "hi", "src_length", "bin"]].to_csv(
    OUT / "test_metadata.csv", index=False
)

print("Total usable sentence pairs:", len(df))
print("Train:", len(train_df))
print("Validation:", len(valid_df))
print("Test:", len(test_df))
print("\nCounts by bin:")
print(pd.crosstab(df["bin"], columns="all"))
print("\nSplit distribution:")
print(pd.crosstab(
    pd.concat([
        train_df.assign(split="train"),
        valid_df.assign(split="valid"),
        test_df.assign(split="test")
    ])["bin"],
    pd.concat([
        train_df.assign(split="train"),
        valid_df.assign(split="valid"),
        test_df.assign(split="test")
    ])["split"]
))

# Verify source and target line alignment.
for name in ["train", "valid", "test"]:
    src_lines = (OUT / f"{name}.en").read_text(
        encoding="utf-8"
    ).splitlines()
    tgt_lines = (OUT / f"{name}.hi").read_text(
        encoding="utf-8"
    ).splitlines()

    assert len(src_lines) == len(tgt_lines)
    print(f"{name}: {len(src_lines)} aligned pairs")

print("\nCreated files:")
for p in sorted(OUT.iterdir()):
    print(p)
```

```python

from pathlib import Path
import re

data_dir = Path("mt_data")

for split in ["train", "valid", "test"]:
    for lang in ["en", "hi"]:
        path = data_dir / f"{split}.{lang}"

        # Preserve one sentence per line while normalizing
        # unusual spaces and repeated whitespace.
        with path.open("r", encoding="utf-8") as f:
            lines = f.read().splitlines()

        cleaned = []
        for line in lines:
            line = line.replace("\u00a0", " ")
            line = line.replace("\u202f", " ")
            line = line.replace("\u2007", " ")
            line = re.sub(r"[ \t]+", " ", line).strip()
            cleaned.append(line)

        with path.open("w", encoding="utf-8") as f:
            f.write("\n".join(cleaned) + "\n")

        print(f"Normalized {path}: {len(cleaned)} lines")
```
